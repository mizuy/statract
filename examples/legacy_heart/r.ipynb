# R Analysis Notebook

This notebook demonstrates how to load preprocessed data from `cache/build` using `load_parquet_dir` and perform analysis.

## Workflow
1. Run `python build.py` to preprocess data and save to `cache/build/`
2. Load data in this notebook using `load_parquet_dir`
3. Perform analysis (Table One, survival analysis, etc.)

```
import sys
from pathlib import Path

_examples = Path.cwd().resolve().parent
if (_examples / "support.py").exists() and str(_examples) not in sys.path:
    sys.path.insert(0, str(_examples))

import polars as pl
from support import load_parquet_dir
from statract import tableone

from project import project
```

## Load Data

Load preprocessed data from `cache/build/` directory.

```
# Load all parquet files from cache/build
data = load_parquet_dir(project.cache / "build")

# Access individual DataFrames
target = data["target"]
print(f"Loaded {len(data)} DataFrames")
print(f"target: {target.shape}")
```

```
# Preview the data
target.head()
```

## Table One

Generate baseline characteristics table.

```
# Define Table One parameters
params = {
    "Age (normalized)": ("age_normalized", "mean_sd"),
    "Year": ("year", "mean_sd"),
    "Surgery": ("surgery", "category"),
    "Transplant": ("transplant", "category"),
    "Event": ("event_bool", "ratio_np"),
    "Survival Time": ("survival_time", "mean_sd"),
}

# Generate Table One
table1 = tableone(target, params)
table1
```
